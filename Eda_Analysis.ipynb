```python
import numpy as np 
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("Iris.csv") 
df.head(3)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.Species.value_counts()
```

```python
df = df.drop(columns=["Id"],axis=1)
df.head()
```

```python
df.describe()
```

```python
df["SepalLengthRatio"] = df.SepalLengthCm / df.SepalWidthCm 
df["PetalLengthRatio"] = df.PetalLengthCm / df.PetalWidthCm  

df["SepalElipseAreaCm2"] =  (np.pi * (df.SepalLengthCm / 2) * (df.SepalWidthCm / 2))
df["PetalElipseAreaCm2"] =  (np.pi * (df.PetalLengthCm / 2) * (df.PetalLengthCm / 2))

df.head()
```

```python
X = df.drop(columns=["Species"],axis=1) 
Y = df.Species
```

```python
corr_matrix = X.corr()
sns.heatmap(data=corr_matrix,fmt='1.3f',annot=True,cmap='inferno')
plt.title("İris Dataset Numerical Features Correlation Matrix")
```

```python
sns.pairplot(df,height=1.6)
```

```python
(fig,ax) = plt.subplots(nrows=2,ncols=4,figsize=(11,6))

sns.violinplot(data=df,y="SepalLengthCm",ax=ax[0,0])
sns.violinplot(data=df,y="SepalWidthCm",ax=ax[0,1])  
sns.violinplot(data=df,y="PetalLengthCm",ax=ax[0,2]) 
sns.violinplot(data=df,y="PetalWidthCm",ax=ax[0,3])
sns.violinplot(data=df,y="SepalLengthRatio",ax=ax[1,0])
sns.violinplot(data=df,y="PetalLengthRatio",ax=ax[1,1]) 
sns.violinplot(data=df,y="SepalElipseAreaCm2",ax=ax[1,2]) 
sns.violinplot(data=df,y="PetalElipseAreaCm2",ax=ax[1,3]) 

plt.suptitle("İris Datasets all of the violin plots")
plt.tight_layout()
```
